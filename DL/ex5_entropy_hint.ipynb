# Hints for entropy and mutual information


Entropy is a measure of uncertainty or randomness in a dataset. In the context of images, it quantifies the amount of information contained in the image. A higher entropy value indicates a more complex image with more details, while a lower entropy value suggests a simpler image with less information.

Mutual information, on the other hand, measures the amount of information that two images share. It quantifies the reduction in uncertainty about one image given knowledge of the other image. High mutual information indicates that the images are similar or contain related information, while low mutual information suggests that the images are different or contain independent information (you can think of mutual information as a measure of similarity between two images).

```python
# These are library aided solutions that will work for the image entropy and mutual information bit.
import numpy as np
from skimage.filters.rank import entropy
from skimage.morphology import disk
from sklearn.metrics import mutual_info_score

def image_entropy(image: np.ndarray) -> float:
    """ Compute the entropy of an image """
    return entropy((image*255).astype(np.uint8), disk(5)).mean()

def image_mutual_information(a: np.ndarray, b: np.ndarray) -> float:
    """ Compute the mutual information between two images a and b"""
    a = (a*255).astype(np.uint8).ravel()
    b = (b*255).astype(np.uint8).ravel()
    return mutual_info_score(a, b)
```
