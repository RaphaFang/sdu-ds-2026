```python
# import essential modules

import nltk
from nltk import word_tokenize
from nltk.corpus import stopwords
nltk.download('punkt_tab')   # required for the tokenizer
nltk.download('stopwords')

import pandas as pd
```

# Word Embeddings

With the following code a Word2Vec model is created from the corpus:

```python
# Load Text
with open("danish_news_headlines.txt", "r") as f:
  lines = f.readlines()

def preprocess(tokens):
  stops = stopwords.words('danish')
  punctuations = [".", ",", "(", ")", "?", "!", "#","'", '"', "[", "]"]
  stops.extend(punctuations)
  return [token for token in tokens if token not in stops]

texts = [preprocess(word_tokenize(line)) for line in lines]


print(len(texts))
```

```python
import gensim

m = gensim.models.Word2Vec(texts, vector_size=100, window=5, min_count=0, workers=4)
```
